```python
import subprocess
import time
import requests

# Stop any old server on port 8000
subprocess.run(
    ["fuser", "-k", "8000/tcp"],
    capture_output=True
)

# Start the updated server
log_file = open("/content/earthflip.log", "w")

server = subprocess.Popen(
    ["python", "server.py"],
    stdout=log_file,
    stderr=subprocess.STDOUT
)

time.sleep(5)

try:
    response = requests.get(
        "http://127.0.0.1:8000/",
        timeout=10
    )
    print("Server Status:", response.status_code)
except Exception as e:
    print("Server Error:", e)
    print(open("/content/earthflip.log").read()[-3000:])
```

```python
import requests
import json
import os

folder = "/content/drive/MyDrive/Colab Notebooks/NASA"

url = "http://127.0.0.1:8000/api/analyze"

params = {
    "start_year": 1981,
    "end_year": 2025,
    "fdr_alpha": 0.05
}

response = requests.get(
    url,
    params=params,
    timeout=300
)

print("API Status:", response.status_code)
response.raise_for_status()

data = response.json()

file_path = os.path.join(
    folder,
    "earthflip_data.json"
)

with open(file_path, "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False)

print("JSON saved successfully!")
print("File:", file_path)
print("Regions:", len(data.get("regions", [])))
```
